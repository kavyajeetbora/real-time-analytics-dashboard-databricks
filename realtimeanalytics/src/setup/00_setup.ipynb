```python
CATALOG = dbutils.widgets.get('CATALOG')
SCHEMA = dbutils.widgets.get('SCHEMA')
```

## Set Catalog

```python
spark.sql(f"USE CATALOG {CATALOG}")
```

## Create Schema

Where we will be storing the tables

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {SCHEMA}")
print(f"Schema: {SCHEMA} is created")
```

## Create Volumes

Where we will be storing the 
1. `landing` folder: for receiving real-time data from a producer
2. `checkpoints` folder

```python
LANDING = dbutils.widgets.get('LANDING')
CHECKPOINTS = dbutils.widgets.get('CHECKPOINTS')
```

```python
spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOG}.{SCHEMA}.{CHECKPOINTS}")
print(f"Volume for checkpoints ready")

spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOG}.{SCHEMA}.{LANDING}")
print(f"Volume for landing ready")
```
